Napisz program, który określi, ile kolejnych liter alfabetu znajduje się między dwiema podanymi.

Wejście
Dwie różne małe litery języka łacińskiego. Pierwsza litera znajduje się wcześniej w alfabecie od drugiej.

Wyjście
Jedna liczba określająca ilość liter.

zastosuj funkcje ord(), np.: ord(a)

```python
a = input("Podaj pierwszą literę: ")
b = input("Podaj drugą literę: ")

wynik = ord(b) - ord(a) - 1

print("Liczba liter pomiędzy:", wynik)
```

Film rozpoczyna się o podanej godzinie i minucie. Pobrany zostaje również czas trwania filmu w minutach.

Oblicz, o której godzinie film się zakończy.
Przykład:

Godzina rozpoczęcia: 21
Minuta rozpoczęcia: 45
Czas filmu: 140

Film zakończy się o: 23:05

```python
godzina = int(input("Godzina rozpoczęcia: "))
minuta = int(input("Minuta rozpoczęcia: "))
czas = int(input("Czas filmu w minutach: "))

# Zamieniamy godzinę rozpoczęcia na minuty
suma_minut = godzina * 60 + minuta + czas

# Obliczamy godzinę i minuty zakończenia
godzina_koniec = (suma_minut // 60) % 24
minuta_koniec = suma_minut % 60

print("Film zakończy się o:", godzina_koniec, ":", minuta_koniec)
```
