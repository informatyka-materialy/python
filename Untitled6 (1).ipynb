```python
def najwieksza(a, b, c):
    naj = a

    if b > naj:
        naj = b

    if c > naj:
        naj = c

    return naj


# Program główny
a = float(input("Podaj pierwszą liczbę: "))
b = float(input("Podaj drugą liczbę: "))
c = float(input("Podaj trzecią liczbę: "))

wynik = najwieksza(a, b, c)

print("Największa liczba to:", wynik)
```

Napisz funkcję oblicz_srednia(oceny), która przyjmuje listę ocen i zwraca ich średnią arytmetyczną. Następnie napisz drugą funkcję czy_zdal(oceny), która wykorzystuje pierwszą funkcję i zwraca informację, czy średnia wynosi co najmniej 2.0.

```python
def oblicz_srednia(oceny):
    suma = 0

    for ocena in oceny:
        suma += ocena

    srednia = suma / len(oceny)
    return srednia


def czy_zdal(oceny):
    srednia = oblicz_srednia(oceny)

    if srednia >= 2.0:
        return True
    else:
        return False


# Program główny
oceny = [3, 4, 2, 5, 3]

wynik = oblicz_srednia(oceny)

print("Średnia ocen:", wynik)

if czy_zdal(oceny):
    print("Uczeń zdał.")
else:
    print("Uczeń nie zdał.")
```

Napisz prosty kalkulator wykorzystujący cztery funkcje:
- dodaj(a, b) – dodawanie,
- odejmij(a, b) – odejmowanie,
- pomnoz(a, b) – mnożenie,
- podziel(a, b) – dzielenie.
Program powinien pobierać dwie liczby oraz znak działania od użytkownika, a następnie wywoływać odpowiednią funkcję. Uwzględnij próbę dzielenia przez zero.

```python
def dodaj(a, b):
    return a + b


def odejmij(a, b):
    return a - b


def pomnoz(a, b):
    return a * b


def podziel(a, b):
    if b == 0:
        return "Nie można dzielić przez zero!"
    else:
        return a / b


# Program główny
a = float(input("Podaj pierwszą liczbę: "))
b = float(input("Podaj drugą liczbę: "))

dzialanie = input("Wybierz działanie (+, -, *, /): ")

if dzialanie == "+":
    wynik = dodaj(a, b)
    print("Wynik:", wynik)

elif dzialanie == "-":
    wynik = odejmij(a, b)
    print("Wynik:", wynik)

elif dzialanie == "*":
    wynik = pomnoz(a, b)
    print("Wynik:", wynik)

elif dzialanie == "/":
    wynik = podziel(a, b)
    print("Wynik:", wynik)

else:
    print("Nieprawidłowy znak działania!")
```
